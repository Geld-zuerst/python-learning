```python
# Palindrome checker
def is_palindrome(value):
    value = "".join(
        character.lower()
        for character in value
        if character.isalnum()
    )

    return value == value[::-1]


print(is_palindrome("Madam"))
print(is_palindrome("Python"))
```

```python
# Vowel counter
def count_vowels(value):
    vowels = "aeiou"
    count = 0

    for character in value.lower():
        if character in vowels:
            count += 1

    return count


print(count_vowels("Python Programming"))
```

```python
# Character frequency
def character_frequency(value):
    frequency = {}

    for character in value:
        frequency[character] = frequency.get(character, 0) + 1

    return frequency


print(character_frequency("banana"))
```

```python
# Caesar cipher
def caesar(text, shift):
    result = ""

    for character in text:
        if character.isalpha():
            start = ord("A") if character.isupper() else ord("a")
            new_character = chr(
                (ord(character) - start + shift) % 26 + start
            )
            result += new_character
        else:
            result += character

    return result


encrypted = caesar("Hello, World!", 3)

print(encrypted)
print(caesar(encrypted, -3))
```

```python
# Anagram checker
def is_anagram(first, second):
    first = first.replace(" ", "").lower()
    second = second.replace(" ", "").lower()

    return sorted(first) == sorted(second)


print(is_anagram("Listen", "Silent"))
print(is_anagram("Python", "Java"))
```

```python
# Reverse every word
def reverse_words(sentence):
    return " ".join(word[::-1] for word in sentence.split())


print(reverse_words("hello world"))
```

```python
# Run-length encoding
def compress(text):
    if not text:
        return ""

    result = ""
    count = 1

    for index in range(1, len(text)):
        if text[index] == text[index - 1]:
            count += 1
        else:
            result += text[index - 1] + str(count)
            count = 1

    return result + text[-1] + str(count)


print(compress("aaabbc"))
```

```python
# First non-repeating character
def first_unique(text):
    for character in text:
        if text.count(character) == 1:
            return character

    return "none"


print(first_unique("swiss"))
print(first_unique("aabb"))
```

```python
# OTP validation
def valid_otp(otp):
    return len(otp) == 6 and otp.isdecimal()


print(valid_otp("123456"))
print(valid_otp("12345"))
print(valid_otp("12a456"))
```

```python
# Name initials
def initials(name):
    words = name.split()
    return ".".join(word[0].upper() for word in words) + "."


print(initials("asha kumari verma"))
```

```python
# File extension
def file_extension(filename):
    before, dot, extension = filename.rpartition(".")

    if not dot:
        return "No extension"

    return extension


print(file_extension("photo.final.jpg"))
print(file_extension("README"))
```
