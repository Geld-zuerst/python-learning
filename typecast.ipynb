### Typecasting (type conversion) is the process of converting a value from one data type to another.


# Implicit typecasting: Python automatically converts a smaller type to a larger compatible type
a = 10
b = 2.5
result = a + b
print("Implicit typecasting example:", result, type(result))  # float

# Explicit typecasting: manually converting a value using built-in functions
x = "100"

y = int(x)
z = float(x)
w = bool(x)

print("Explicit conversion examples:")
print("int('100') ->", y, type(y))
print("float('100') ->", z, type(z))
print("bool('100') ->", w, type(w))

# Additional examples
print("str(42) ->", str(42), type(str(42)))
print("int(7.8) ->", int(7.8))
Typecasting is the process of converting a value from one data type to another so that it can be used correctly in a particular operation or context.

In Python, the main representative data types are:
- int: whole numbers
- float: decimal numbers
- str: text
- bool: true/false values
- complex: numbers with a real and imaginary part

There are two main types of typecasting:

1. Implicit typecasting
This happens automatically when Python converts a smaller, compatible data type into a larger one without losing meaning. For example, when an integer is used with a floating-point number, Python usually converts the integer to a float so the expression can be evaluated correctly. This is done automatically by the interpreter.

2. Explicit typecasting
This is done manually by the programmer when they want to convert a value to a specific type. It is useful when a value is in the wrong format for a required operation, such as converting a numeric string into an integer or float, or converting a number into a boolean value.

Typecasting is important because different operations require different data types. It helps in arithmetic operations, data handling, input processing, and making sure variables are compatible with the functions or expressions they are used in. Without typecasting, operations involving different data types may fail or produce incorrect results.
