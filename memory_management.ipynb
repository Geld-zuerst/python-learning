# Three Pillars of Python Memory Management

Python memory management relies on three main pillars:

1. **Reference Counting**  
    Python tracks how many references point to an object. When an object's reference count reaches zero, its memory is released immediately.

2. **Garbage Collection**  
    The cyclic garbage collector identifies and removes unreachable objects that reference each other. This is especially useful for objects that reference counting cannot release alone.

3. **Memory Allocation**  
    Python's memory manager allocates and reuses memory efficiently. It organizes memory into blocks and pools, while the operating system manages larger memory allocations.

# Garbage Collection in Python

Python manages memory primarily through **reference counting**. When an object’s reference count reaches zero, it can be deallocated. Reference counting alone cannot reclaim objects that refer to one another in a cycle.

Python’s **cyclic garbage collector** detects and cleans up many such unreachable reference cycles. The `gc` module lets you inspect and control this collector:

```python
import gc

print(gc.isenabled())  # Check whether cyclic garbage collection is enabled
gc.collect()           # Request a garbage-collection run
```

Garbage collection is usually automatic; manually calling `gc.collect()` is rarely necessary.

# Python Reference Counting

Python uses **reference counting** to manage memory. An object's reference count increases when it is assigned or passed to a new variable, and decreases when references are deleted or go out of scope.

A `refcount` of `0` means the object is no longer referenced and can be deallocated.


````python
x = [1,2,3,4,5,6]
print(sys.getrefcount(x))
````

````python
y=x
print(sys.getrefcount(x))
````

## Cell 16
```python
x = [1, 2, 3, 4, 5, 6]
print(sys.getrefcount(x))
```

Creates a list and stores it in `x`. `sys.getrefcount()` reports the list's reference count, including the temporary reference passed to the function:

```text
2
```

## Cell 17
```python
y = x
print(sys.getrefcount(x))
```

`y` refers to the same list as `x`, so it adds another reference:

```text
3
```

## Cell 18
```python
ab = x
print(sys.getrefcount(x))
```

`ab` is another name for the same list. It adds yet another reference:

```text
4
```

All variables `x`, `y`, and `ab` refer to one shared list object. Adding aliases increases its reference count, not the number of list objects.

In CPython, `sys.getrefcount()` itself temporarily adds a reference, so the returned count is one higher than the number of ordinary references.

````python
p =[1,2,3,4]
q =[5,6,7,8]

p.append(q)
q.append(p)
print(sys.getrefcount(p))
print(sys.getrefcount(q))

del(p)
print(sys.getrefcount(q))
````
