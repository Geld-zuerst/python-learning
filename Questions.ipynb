```python
x = 10
def int():
    global x
    x=50

int()
print(x)
```

```python
score = 10

def update_score():
    global score
    score = 25

update_score()
print(score)
```

```python
count = 3

def increase_count():
    global count
    count += 1

increase_count()
print(count)
```

```python
message = "Hello"

def update_message():
    global message
    message = "Updated"

update_message()
print(message)
```

```python
x = "Global"

def outer():
    x = "Enclosing"
    def inner():
        x = "Local"
        print(x)
    inner()
outer()
```

```python
def sqr(*args):
    for a in args:
        print(a * a)

sqr(2)
sqr(3)
sqr(4,5,6,7,8,9)
```

```python
def sqr2(a):
    print(a**2)

sqr(2)
sqr(4)
```

```python
import math

def square_root(number):
    return math.sqrt(number)

print(square_root(25))
```
