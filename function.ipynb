```python
def opr(a, b, operation):
    if operation == "+":
        print(a+b)
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "+":
        return a+b
    else:
        print("ERROR")

opr(4,5,"+")
```

```python
opr(98765,567,"+")
```

```python
opr(2345,98765,"+")
```

```python
def show(*args):
    print(sum(args))

show(1,2)
```

```python
def show(*args):
    result = args[0]
    for num in args[1:]:
        result -= num
    print(result)
```

```python
show(2,1)
```

```python
def show(**k):
    print(k)
    print(type(k))

show(name="Harsh", age=19)
```

```python
def test(*args,**kwargs):
    print("ARGS:", args)
    print("KWARGS:", kwargs)

test(10,20,name='krish',age=20)
```
