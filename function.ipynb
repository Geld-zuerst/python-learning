```python
def opr(a, b, operation):
    if operation == "+":
        print(a+b)
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "+":
        return a+b
    else:
        print("ERROR")

opr(4,5,"+")
```

```python
opr(98765,567,"+")
```

```python
opr(2345,98765,"+")
```
